# Selection

Use the **bao** kernel in your allocated HPC remote desktop. Edit these cells and the reusable functions in `pzbao/diagnostics/selection.py`.

These small synthetic examples illustrate the interface; they are not BAO results. Scientific interpretation and acceptance remain yours.


```python
from pathlib import Path
import numpy as np
from pzbao.diagnostics import selection as diagnostics
from pzbao.config import load_analysis_config
from pzbao.config.stages import stage_document
from pzbao.run import load_run, stage_units

RUN_ROOT = None  # Set to the named run directory you want to inspect.
run = load_run(RUN_ROOT) if RUN_ROOT is not None else None
config = run["config"] if run is not None else load_analysis_config()
```

## Settings and saved products

This is the section used by `bao select`. Edit a complete stage YAML or a combined experiment YAML. CLI runs freeze scientific settings; use a new named run for a changed experiment.


```python
settings = stage_document(config, "select")
settings
```

```python
if run is not None:
    units = stage_units("select", config, run["metadata"]["workload"])
    products = list((run["root"] / "select" / "units").glob("*/result.yaml"))
    print(f"{len(products)} saved units; {len(units)} configured units")
```

## Explore and plot

Replace the illustrative arrays with saved arrays from your chosen unit. Plotting functions accept an existing axis and return their data for inspection.


```python
artifact = {
    "z_edges": np.linspace(0.4, 1.0, 5),
    "mean_counts": np.array([10.0, 20.0, 20.0, 10.0]),
    "target_probability_mass": np.array([0.2, 0.3, 0.3, 0.2]),
}
ax, values = diagnostics.plot_radial_target(artifact)
values
```

## Load one saved result

Set `LOAD_SAVED = True` after selecting your run and unit. These cells only read saved products; missing prerequisites remain visible as errors. The clustering example reads xi; P(k)-only runs can be inspected through the unit's `measurements.json` and named power NPZ files.


```python
LOAD_SAVED = False
SAVED_UNIT_INDEX = 0  # Choose the configured unit/route to inspect.
if LOAD_SAVED:
    if run is None:
        raise ValueError("Set RUN_ROOT before loading saved products")
    from pzbao.selection import read_preparation

    preparation = read_preparation(run, units[SAVED_UNIT_INDEX]["tracer"])
    CONDITION = config["analysis"]["workloads"][run["metadata"]["workload"]]["sigma_z_tiers"][0]
    CAP = "NGC"
    radial = preparation["conditions"][CONDITION][CAP]["radial"]
    saved_ax, saved_values = diagnostics.plot_radial_target(radial)
```
